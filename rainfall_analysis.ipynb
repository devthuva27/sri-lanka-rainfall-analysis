```python
import pandas as pd

df = pd.read_csv("SriLanka_Weather_Dataset.csv")
```

```python
df.info()
```

```python
df['time'] = pd.to_datetime(df['time'])
```

```python
df = df[['time', 'city', 'rain_sum']]
```

```python
df = df.rename(columns={
    'time': 'date',
    'rain_sum': 'rainfall'
})
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df['year'] = df['date'].dt.year
```

```python
df.head()
```

```python
df['month'] = df['date'].dt.month
```

```python
df.head()
```

```python
df.head()
```

```python
monthly_avg_rainfall = df.groupby('month')['rainfall'].mean()
```

```python
monthly_avg_rainfall
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
monthly_avg_rainfall.plot(kind='bar')
plt.title("Average Monthly Rainfall in Sri Lanka")
plt.xlabel("Month")
plt.ylabel("Average Rainfall")
plt.show()
```

```python
yearly_avg_rainfall = df.groupby('year')['rainfall'].mean() 
```

```python
yearly_avg_rainfall
```

```python
plt.figure(figsize=(10,5))
yearly_avg_rainfall.plot()
plt.title("Average Yearly Rainfall Trend in Sri Lanka")
plt.xlabel("Year")
plt.ylabel("Average Rainfall")
plt.show()
```
